```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# List of model result files
model_files = [
    '../results/outputs/knn_results_IT24102298.csv',
    '../results/outputs/linear_regression_results_IT24102297.csv',
    '../results/outputs/ridge_results_IT24102225.csv',
    '../results/outputs/lasso_results_IT24102308.csv',
    '../results/outputs/random_forest_results_IT24102257.csv',
    '../results/outputs/svr_results_IT24102306.csv'
]

# Initialize an empty list to store results
all_results = []

# Load and process each model's results
for file in model_files:
    df = pd.read_csv(file)
    # For Lasso, convert metrics to match other models' format
    if 'lasso' in file:
        df['Model'] = 'Lasso (IT24102308)'
        df['MSE'] = df['RMSE (Test)'] ** 2
        df['R2'] = df['R² (Test)']
        df['CV R2 Mean'] = df['CV R²']
        df['CV R2 Std'] = 0  # Not provided in Lasso results, set to 0
        df['Runtime (s)'] = None  # Not provided in Lasso results
        df = df[['Model', 'MSE', 'R2', 'CV R2 Mean', 'CV R2 Std', 'Best Params', 'Runtime (s)']]
    all_results.append(df)

# Concatenate all results
results_df = pd.concat(all_results, ignore_index=True)

# Clean up the DataFrame
results_df = results_df.sort_values('R2', ascending=False)
results_df = results_df.round(4)  # Round numerical values to 4 decimal places

# Save comparison results
results_df.to_csv('../results/outputs/model_comparison_results.csv', index=False)
print("Saved comparison results to '../results/outputs/model_comparison_results.csv'")

# Visualization: Bar plot for R2 scores
plt.figure(figsize=(10, 6))
sns.barplot(x='R2', y='Model', data=results_df, palette='viridis')
plt.title('Model Comparison: R² Scores')
plt.xlabel('R² Score')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/model_comparison_r2.png')
plt.close()
print("Saved R² comparison plot to '../results/eda_visualizations/model_comparison_r2.png'")

# Visualization: Bar plot for MSE
plt.figure(figsize=(10, 6))
sns.barplot(x='MSE', y='Model', data=results_df, palette='magma')
plt.title('Model Comparison: Mean Squared Error (MSE)')
plt.xlabel('Mean Squared Error')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/model_comparison_mse.png')
plt.close()
print("Saved MSE comparison plot to '../results/eda_visualizations/model_comparison_mse.png'")

# Visualization: Bar plot for CV R2 Mean
plt.figure(figsize=(10, 6))
sns.barplot(x='CV R2 Mean', y='Model', data=results_df, palette='coolwarm')
plt.title('Model Comparison: Cross-Validation R² Mean')
plt.xlabel('CV R² Mean')
plt.tight_layout()
plt.savefig('../results/eda_visualizations/model_comparison_cv_r2.png')
plt.close()
print("Saved CV R² comparison plot to '../results/eda_visualizations/model_comparison_cv_r2.png'")

# Print summary table
print('\nModel Comparison Summary:')
print(results_df[['Model', 'MSE', 'R2', 'CV R2 Mean', 'CV R2 Std', 'Runtime (s)', 'Best Params']])
```
